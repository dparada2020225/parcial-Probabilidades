# Problema B – Monedas
**MM3014 Teoría de Probabilidades | Parcial 4**

**Experimento base:** Lanzar tres monedas justas (cara = 1, cruz = 0).

- **a)** Estimar P(exactamente 2 caras)
- **b)** Estimar E[X], donde X = número de caras

> **Concepto:** El valor esperado E[X] se estima como el promedio de X
> sobre todas las repeticiones.

```
import random

# -----------------------------------------
# PARAMETROS CONFIGURABLES
# -----------------------------------------
SEMILLA        = 2026
N_REPETICIONES = 10_000
N_MONEDAS      = 3   # Cantidad de monedas
CARAS_EXACTAS  = 2   # Numero exacto de caras para el inciso a
```

```
# -----------------------------------------
# FUNCIONES DEL EXPERIMENTO
# -----------------------------------------

def lanzar_moneda():
    """Retorna 1 (cara) o 0 (cruz) con igual probabilidad."""
    return random.randint(0, 1)

def lanzar_monedas():
    """
    Lanza N_MONEDAS monedas independientes.
    Retorna el total de caras obtenidas.
    """
    return sum(lanzar_moneda() for _ in range(N_MONEDAS))
```

```
# -----------------------------------------
# SIMULACION
# -----------------------------------------
random.seed(SEMILLA)

conteo_caras_exactas = 0  # veces con exactamente CARAS_EXACTAS caras
suma_total_caras     = 0  # acumulado de X para calcular E[X]

for _ in range(N_REPETICIONES):
    x = lanzar_monedas()

    if x == CARAS_EXACTAS:
        conteo_caras_exactas += 1

    suma_total_caras += x

p_caras = conteo_caras_exactas / N_REPETICIONES
e_x     = suma_total_caras / N_REPETICIONES

print(f"P(exactamente {CARAS_EXACTAS} caras) = {p_caras:.4f}")
print(f"E[X] = {e_x:.4f}")
```
