```python
import random

SEMILLA        = 2026
N_REPETICIONES = 10_000
CAJA = ["roja"] * 5 + ["azul"] * 3 + ["verde"] * 2
```

---
## Problema 1 – Canicas (condicional)

```python

N_EXTRAER   = 2
COND_COLOR  = "roja"
EVT_COLOR   = "azul"
EVT_CANT    = 1

random.seed(SEMILLA)
conteo_B = conteo_AB = 0

for _ in range(N_REPETICIONES):
    ext = random.sample(CAJA, N_EXTRAER)
    cond = ext.count(COND_COLOR) >= 1
    evt  = ext.count(EVT_COLOR) == EVT_CANT

    if cond:
        conteo_B += 1
        if evt:
            conteo_AB += 1

p1 = conteo_AB / conteo_B if conteo_B > 0 else 0
print(f"P(exactamente una {EVT_COLOR} | al menos una {COND_COLOR}) = {p1:.4f}")
```

---
## Problema 2 – Monedas (valor esperado)

```python
N_MONEDAS = 3 

random.seed(SEMILLA)
suma_dif = 0

for _ in range(N_REPETICIONES):
    caras  = sum(random.randint(0, 1) for _ in range(N_MONEDAS))
    sellos = N_MONEDAS - caras
    suma_dif += (caras - sellos)

e2 = suma_dif / N_REPETICIONES
print(f"E[caras - sellos] = {e2:.4f}")
```

---
## Problema 3 – Monedas (Bayes)

```python
MONEDAS = ["justa", "justa", "dos_caras"]

def lanzar_moneda(tipo):
    """
    justa     -> cara con prob 0.5
    dos_caras -> cara siempre (prob 1.0)
    """
    if tipo == "dos_caras":
        return 1
    return random.randint(0, 1)

random.seed(SEMILLA)
conteo_cara = conteo_dc_cara = 0

for _ in range(N_REPETICIONES):
    moneda = random.choice(MONEDAS)
    res    = lanzar_moneda(moneda)

    if res == 1:           # salio cara (observacion)
        conteo_cara += 1
        if moneda == "dos_caras":
            conteo_dc_cara += 1

p3 = conteo_dc_cara / conteo_cara if conteo_cara > 0 else 0
print(f"P(moneda de dos caras | salio cara) = {p3:.4f}")
```

---
## Problema 4 – Canicas (todas distintas)

```python
N_EXTRAER_4      = 3
COLORES_OBJETIVO = ["roja", "azul", "verde"]

random.seed(SEMILLA)
conteo_distintos = 0

for _ in range(N_REPETICIONES):
    ext = random.sample(CAJA, N_EXTRAER_4)
    todas_distintas = all(ext.count(c) == 1 for c in COLORES_OBJETIVO)

    if todas_distintas:
        conteo_distintos += 1

p4 = conteo_distintos / N_REPETICIONES
print(f"P(todas de distinto color) = {p4:.4f}")
```

---
## Problema 5 – Monedas (condicional)

```python

N_MONEDAS_5   = 4   
MIN_CARAS     = 2   
CARAS_EXACTAS = 2   

random.seed(SEMILLA)
conteo_B = conteo_AB = 0

for _ in range(N_REPETICIONES):
    caras = sum(random.randint(0, 1) for _ in range(N_MONEDAS_5))

    if caras >= MIN_CARAS:
        conteo_B += 1
        if caras == CARAS_EXACTAS:
            conteo_AB += 1

p5 = conteo_AB / conteo_B if conteo_B > 0 else 0
print(f"P(exactamente {CARAS_EXACTAS} caras | caras >= {MIN_CARAS}) = {p5:.4f}")
```
