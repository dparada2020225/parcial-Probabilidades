# Problema A – Dados
**MM3014 Teoría de Probabilidades | Parcial 4**

**Experimento base:** Lanzar dos dados justos de 6 caras.

- **a)** Estimar P(suma = 7)
- **b)** Estimar P(suma = 7 | al menos uno de los dados es par)

> **Concepto:** La probabilidad condicional P(A|B) se estima filtrando solo los
> experimentos donde B ocurrió, y contando cuántos de esos también cumplen A.

```
import random

# -----------------------------------------
# PARAMETROS CONFIGURABLES
# Cambia estos valores segun el temario
# -----------------------------------------
SEMILLA        = 2026   # Semilla para reproducibilidad
N_REPETICIONES = 10_000 # Numero de experimentos
CARAS_DADO     = 6      # Caras de cada dado
SUMA_OBJETIVO  = 7      # Suma que queremos estimar
```

```
# -----------------------------------------
# FUNCIONES DEL EXPERIMENTO
# -----------------------------------------

def lanzar_dado():
    """Simula un dado justo con CARAS_DADO caras."""
    return random.randint(1, CARAS_DADO)

def lanzar_dos_dados():
    """Lanza dos dados independientes. Retorna (dado1, dado2)."""
    return lanzar_dado(), lanzar_dado()

def condicion_b(dado1, dado2):
    """
    Condicion del evento B para el inciso b.
    ACTUAL: al menos uno de los dados es par.
    Para cambiar: modifica esta funcion.
      Ej. 'al menos uno impar': return (dado1 % 2 != 0) or (dado2 % 2 != 0)
    """
    return (dado1 % 2 == 0) or (dado2 % 2 == 0)
```

```
# -----------------------------------------
# SIMULACION
# -----------------------------------------
random.seed(SEMILLA)

conteo_suma          = 0  # veces que suma == SUMA_OBJETIVO
conteo_B             = 0  # veces que se cumple condicion_b
conteo_suma_y_B      = 0  # veces que se cumplen ambas (A inter B)

for _ in range(N_REPETICIONES):
    d1, d2 = lanzar_dos_dados()
    suma = d1 + d2

    if suma == SUMA_OBJETIVO:
        conteo_suma += 1

    if condicion_b(d1, d2):
        conteo_B += 1
        if suma == SUMA_OBJETIVO:
            conteo_suma_y_B += 1

# Probabilidades
p_suma = conteo_suma / N_REPETICIONES
p_cond = conteo_suma_y_B / conteo_B if conteo_B > 0 else 0

print(f"P(suma = {SUMA_OBJETIVO}) = {p_suma:.4f}")
print(f"P(suma = {SUMA_OBJETIVO} | al menos un par) = {p_cond:.4f}")
```
