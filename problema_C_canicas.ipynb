# Problema C – Canicas de colores
**MM3014 Teoría de Probabilidades | Parcial 4**

**Experimento base:** Caja con 5 rojas, 3 azules y 2 verdes. Se extraen 2 sin reemplazo.

- **a)** Estimar P(ambas son rojas)
- **b)** [Bayes] Dos cajas. Se elige una al azar, se sacan 2 sin reemplazo y resultan 1 roja y 1 verde.  
  Estimar P(Caja 1 | una roja y una verde)

> **Concepto (Bayes):** Filtramos los experimentos donde ocurrió la observación,
> y dentro de esos contamos cuántos correspondían a Caja 1.

```
import random

# -----------------------------------------
# PARAMETROS CONFIGURABLES
# -----------------------------------------
SEMILLA        = 2026
N_REPETICIONES = 10_000
N_EXTRACCIONES = 2

# Composicion de las cajas (agregar/quitar elementos para cambiar cantidades)
CAJA_1 = ["roja"] * 5 + ["azul"] * 3 + ["verde"] * 2   # 10 canicas
CAJA_2 = ["roja"] * 2 + ["azul"] * 5 + ["verde"] * 3   # 10 canicas
```

```
# -----------------------------------------
# FUNCIONES DEL EXPERIMENTO
# -----------------------------------------

def extraer_sin_reemplazo(caja, n):
    """Extrae n canicas al azar sin reemplazo. Retorna lista de colores."""
    return random.sample(caja, n)

def es_resultado_observado(extracciones):
    """
    Define el resultado observado para el inciso b (Bayes).
    ACTUAL: exactamente 1 roja y 1 verde.
    Para cambiar: modifica esta funcion.
      Ej. '2 azules': return extracciones.count('azul') == 2
    """
    return extracciones.count("roja") == 1 and extracciones.count("verde") == 1
```

```
# -----------------------------------------
# SIMULACION
# -----------------------------------------
random.seed(SEMILLA)

conteo_ambas_rojas       = 0
conteo_obs               = 0  # veces que ocurrio la observacion
conteo_caja1_y_obs       = 0  # veces Caja1 Y observacion

for _ in range(N_REPETICIONES):

    # Inciso a: extraer de CAJA_1
    ext_a = extraer_sin_reemplazo(CAJA_1, N_EXTRACCIONES)
    if ext_a.count("roja") == N_EXTRACCIONES:
        conteo_ambas_rojas += 1

    # Inciso b: elegir caja al azar y extraer
    caja_id = random.choice([1, 2])
    caja    = CAJA_1 if caja_id == 1 else CAJA_2
    ext_b   = extraer_sin_reemplazo(caja, N_EXTRACCIONES)

    if es_resultado_observado(ext_b):
        conteo_obs += 1
        if caja_id == 1:
            conteo_caja1_y_obs += 1

p_ambas_rojas = conteo_ambas_rojas / N_REPETICIONES
p_bayes       = conteo_caja1_y_obs / conteo_obs if conteo_obs > 0 else 0

print(f"P(ambas rojas) = {p_ambas_rojas:.4f}")
print(f"P(Caja 1 | una roja y una verde) = {p_bayes:.4f}")
```
